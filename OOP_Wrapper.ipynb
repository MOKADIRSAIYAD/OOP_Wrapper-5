## 1. Person Class (Base Class)
Basic class jo `name` aur `age` store karti hai. `Employee` isi class se inherit karta hai.

```python
class Person:
    """Base class: sirf name aur age."""

    def __init__(self, name=None, age=None):
        self.name = name
        self.age = age

    def display(self):
        print("Person Details:")
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
```

## 2. Employee Class
- Attributes: `employee_id`, `name`, `age`, `salary`
- `employee_id` aur `salary` **private** hain (encapsulation) — getter/setter se access hote hain
- **Method overloading:** Python me true overloading nahi hoti, isliye default arguments aur `@classmethod` constructors use kiye hain
- **Destructor:** `__del__()` resources free karta hai

```python
class Employee(Person):
    _role = "Employee"

    # ---------- Constructor (method overloading using default args) ----------
    # Employee()                              -> empty employee
    # Employee("Jane", 28)                    -> name + age
    # Employee("Jane", 28, "E123", 50000)     -> full details
    def __init__(self, name=None, age=None, employee_id=None, salary=0.0):
        super().__init__(name, age)
        self.__employee_id = employee_id      # private
        self.__salary = float(salary)         # private

    # ---------- Overloaded constructors (alternate ways) ----------
    @classmethod
    def from_string(cls, text):
        """Employee.from_string('Jane,28,E123,50000')"""
        name, age, emp_id, salary = [x.strip() for x in text.split(",")]
        return cls(name, int(age), emp_id, float(salary))

    @classmethod
    def from_dict(cls, data):
        """Employee.from_dict({'name':..., 'age':..., 'employee_id':..., 'salary':...})"""
        return cls(data.get("name"), data.get("age"),
                   data.get("employee_id"), data.get("salary", 0.0))

    # ---------- Getters & Setters (encapsulation) ----------
    def get_employee_id(self):
        return self.__employee_id

    def set_employee_id(self, employee_id):
        if not str(employee_id).strip():
            raise ValueError("Employee ID cannot be empty.")
        self.__employee_id = employee_id

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if float(salary) < 0:
            raise ValueError("Salary cannot be negative.")
        self.__salary = float(salary)

    # ---------- Display ----------
    def display(self):
        print(f"{self._role} Details:")
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Employee ID: {self.get_employee_id()}")
        print(f"Salary: ${self.get_salary()}")

    # ---------- Destructor ----------
    def __del__(self):
        print(f"[Cleanup] {self._role} '{getattr(self, 'name', None)}' object destroyed, resources freed.")
```

## 3. Manager Class (Derived Class)
`Employee` se inherit karta hai, extra attribute `department` add karta hai aur `display()` ko **override** karta hai.

```python
class Manager(Employee):
    _role = "Manager"

    def __init__(self, name=None, age=None, employee_id=None, salary=0.0, department=None):
        super().__init__(name, age, employee_id, salary)   # parent constructor call
        self.department = department

    def display(self):                                      # method overriding
        super().display()                                   # parent ka display
        print(f"Department: {self.department}")
```

## 4. Developer Class (Derived Class)
`Employee` se inherit karta hai, extra attribute `programming_language` add karta hai aur `display()` ko **override** karta hai.

```python
class Developer(Employee):
    _role = "Developer"

    def __init__(self, name=None, age=None, employee_id=None, salary=0.0, programming_language=None):
        super().__init__(name, age, employee_id, salary)
        self.programming_language = programming_language

    def display(self):                                      # method overriding
        super().display()
        print(f"Programming Language: {self.programming_language}")
```

## 5. Quick Demo (Overloading, Encapsulation, `issubclass()`)

```python
# --- Method overloading demo ---
e1 = Employee()                                   # no arguments
e2 = Employee("Jane Smith", 28)                   # 2 arguments
e3 = Employee("Bob Lee", 35, "E200", 45000)       # all arguments
e4 = Employee.from_string("Carl Young,31,E300,52000")
e5 = Employee.from_dict({"name": "Dina Rao", "age": 26, "employee_id": "E400", "salary": 48000})

for emp in (e1, e2, e3, e4, e5):
    emp.display()
    print("-" * 30)

# --- Encapsulation demo ---
e3.set_salary(60000)
print("Updated salary:", e3.get_salary())
try:
    e3.set_salary(-5)
except ValueError as err:
    print("Error:", err)

# Private attribute directly access nahi hota
try:
    print(e3.__salary)
except AttributeError:
    print("Direct access to __salary is blocked (private attribute).")

# --- issubclass() demo ---
print("\nissubclass(Manager, Employee)   :", issubclass(Manager, Employee))
print("issubclass(Developer, Employee) :", issubclass(Developer, Employee))
print("issubclass(Employee, Person)    :", issubclass(Employee, Person))
print("issubclass(Person, Employee)    :", issubclass(Person, Employee))

# --- Overriding demo ---
print()
Manager("Alice Johnson", 40, "M456", 80000, "Sales").display()
print()
Developer("Rahul Mehta", 29, "D789", 70000, "Python").display()
```

## 6. Menu-Driven User Interface
Menu se user Person / Employee / Manager / Developer bana sakta hai, details dekh sakta hai, salary update kar sakta hai, employee remove kar sakta hai aur program se exit kar sakta hai.

```python
# ------------------ Input helper functions ------------------
def get_text(prompt):
    while True:
        value = input(prompt).strip()
        if value:
            return value
        print("Input cannot be empty. Try again.")

def get_int(prompt):
    while True:
        try:
            value = int(input(prompt))
            if value <= 0:
                print("Please enter a positive number.")
                continue
            return value
        except ValueError:
            print("Invalid input. Please enter a number.")

def get_float(prompt):
    while True:
        try:
            value = float(input(prompt))
            if value < 0:
                print("Value cannot be negative.")
                continue
            return value
        except ValueError:
            print("Invalid input. Please enter a number.")


# ------------------ Storage ------------------
persons, employees, managers, developers = [], [], [], []


# ------------------ Menu actions ------------------
def create_person():
    name = get_text("Enter Name: ")
    age = get_int("Enter Age: ")
    persons.append(Person(name, age))
    print(f"\nPerson created with name: {name} and age: {age}.")

def create_employee():
    name = get_text("Enter Name: ")
    age = get_int("Enter Age: ")
    emp_id = get_text("Enter Employee ID: ")
    salary = get_float("Enter Salary: ")
    employees.append(Employee(name, age, emp_id, salary))
    print(f"\nEmployee created with name: {name}, age: {age}, ID: {emp_id}, and salary: ${salary}.")

def create_manager():
    name = get_text("Enter Name: ")
    age = get_int("Enter Age: ")
    emp_id = get_text("Enter Employee ID: ")
    salary = get_float("Enter Salary: ")
    dept = get_text("Enter Department: ")
    managers.append(Manager(name, age, emp_id, salary, dept))
    print(f"\nManager created with name: {name}, age: {age}, ID: {emp_id}, salary: ${salary}, and department: {dept}.")

def create_developer():
    name = get_text("Enter Name: ")
    age = get_int("Enter Age: ")
    emp_id = get_text("Enter Employee ID: ")
    salary = get_float("Enter Salary: ")
    lang = get_text("Enter Programming Language: ")
    developers.append(Developer(name, age, emp_id, salary, lang))
    print(f"\nDeveloper created with name: {name}, age: {age}, ID: {emp_id}, salary: ${salary}, and language: {lang}.")

def show_details():
    print("\nChoose details to show:")
    print("1. Person\n2. Employee\n3. Manager\n4. Developer")
    choice = input("Enter your choice: ").strip()
    data = {"1": persons, "2": employees, "3": managers, "4": developers}.get(choice)
    if data is None:
        print("Invalid choice.")
    elif not data:
        print("\nNo records found.")
    else:
        for obj in data:
            print()
            obj.display()

def find_employee(emp_id):
    for group in (employees, managers, developers):
        for obj in group:
            if obj.get_employee_id() == emp_id:
                return obj, group
    return None, None

def update_salary():
    emp_id = get_text("Enter Employee ID: ")
    obj, _ = find_employee(emp_id)
    if obj is None:
        print("Employee not found.")
        return
    obj.set_salary(get_float("Enter New Salary: "))
    print(f"Salary updated. New salary: ${obj.get_salary()}")

def remove_employee():
    emp_id = get_text("Enter Employee ID: ")
    obj, group = find_employee(emp_id)
    if obj is None:
        print("Employee not found.")
        return
    group.remove(obj)
    del obj
    print(f"Employee {emp_id} removed.")


# ------------------ Main menu ------------------
def main():
    print("--- Python OOP Project: Employee Management System ---")
    while True:
        print("\nChoose an operation:")
        print("1. Create a Person")
        print("2. Create an Employee")
        print("3. Create a Manager")
        print("4. Create a Developer")
        print("5. Show Details")
        print("6. Update Salary")
        print("7. Remove Employee")
        print("8. Exit")
        choice = input("\nEnter your choice: ").strip()
        print()

        if choice == "1":
            create_person()
        elif choice == "2":
            create_employee()
        elif choice == "3":
            create_manager()
        elif choice == "4":
            create_developer()
        elif choice == "5":
            show_details()
        elif choice == "6":
            update_salary()
        elif choice == "7":
            remove_employee()
        elif choice == "8":
            persons.clear(); employees.clear(); managers.clear(); developers.clear()
            print("Exiting the system. All resources have been freed.")
            print("\nGoodbye!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 8.")
        print("\n--- Choose another operation ---")

main()
```
